## Ollama’yı Başlat


```python
import threading
import subprocess
import time
import os

def run_ollama_serve():
    # Impostiamo le variabili d'ambiente correttamente
    os.environ["OLLAMA_HOST"] = "0.0.0.0"
    os.environ["OLLAMA_PORT"] = "11434"
    subprocess.Popen(["ollama", "serve"])

# Avvia il server in un thread separato
thread = threading.Thread(target=run_ollama_serve)
thread.start()

# Attendi che il server si avvii
time.sleep(5)
print("Ollama server avviato correttamente.")
```

## Ngrok Tunnel
Ollama viene eseguito localmente. Per accedervi esternamente o tramite webhook, utilizzeremo `ngrok` per creare un tunnel pubblico.

```python
!pip install pyngrok
```

```python
from pyngrok import ngrok
from google.colab import userdata
import os

# Prendi il token dai segreti di Colab
NGROK_TOKEN = userdata.get('NGROK_TOKEN')
ngrok.set_auth_token(NGROK_TOKEN)

# Avvia il tunnel sulla porta 11434 (default di Ollama)
public_url = ngrok.connect(11434)
print(f"Il tuo tunnel pubblico è attivo su: {public_url}")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM
from IPython.display import Markdown

template = """Question: {question}

Answer: Let's think step by step."""

prompt = ChatPromptTemplate.from_template(template)
model = OllamaLLM(model="llama3.2")
chain = prompt | model

Markdown(chain.invoke({
    "question": "Dik üçgende hipotenüs nedir?"
}))
```
